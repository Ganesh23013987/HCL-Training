**Question 10 – Student Ranking**

```python
import numpy as np

marks = np.array([245, 278, 219, 290, 256])

sorted_marks = np.sort(marks)[::-1]

ranking = np.argsort(marks)[::-1]

print("Original Marks:", marks)

print("\nMarks in descending order:")
print(sorted_marks)

print("\nStudent ranking:")
for rank, index in enumerate(ranking, start=1):
    print("Rank", rank, "- Student", index + 1,
          "- Marks:", marks[index])
```

**Question 15 – Student Performance Analysis**

```python
import numpy as np

marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])

print("Student Marks:")
print(marks)

total = np.sum(marks, axis=1)

average = np.mean(marks, axis=1)

highest = np.max(marks, axis=1)

lowest = np.min(marks, axis=1)

class_average = np.mean(marks)

above_average = total / 3 > class_average

print("\nTotal Marks:")
print(total)

print("\nAverage Marks:")
print(average)

print("\nHighest Marks:")
print(highest)

print("\nLowest Marks:")
print(lowest)

print("\nClass Average:", class_average)

print("\nStudents above class average:")

for i in range(5):
    if above_average[i]:
        print("Student", i + 1)
```

# **Python Assignments:**


```python
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

print("Prime numbers:")

for num in range(start, end + 1):

    if num < 2:
        continue

    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

```python
def factorial(n):

    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


n = int(input("Enter a number: "))

result = factorial(n)

print("Factorial:", result)
```

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

n = int(input("Enter the number: "))

print("Fibonacci series: ")
for i in range(n):
    print(fib(i), end=" ")
```

# **Python Function Assignments**

```python
def calculate(a, b, operation):

    if operation == "+":
        return a + b

    elif operation == "-":
        return a - b

    elif operation == "*":
        return a * b

    elif operation == "/":
        return a / b

    else:
        return "Invalid operation"


a = int(input())
b = int(input())
operation = input()

print(calculate(a, b, operation))
```

```python
def sum_numbers(*args):

    total = 0

    for n in args:
        total = total + n

    return total

nums=list(map(int,input().split()))
print(sum_numbers(*nums))
```

```python
def remove_duplicates(lst):

    result = []

    for n in lst:
        if n not in result:
            result.append(n)

    return result


numbers = list(map(int,input().split()))

print(remove_duplicates(numbers))
```
