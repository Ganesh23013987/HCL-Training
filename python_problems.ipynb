```python
def longest_attendance(ids):
    seen = []
    max_len = 0

    for i in range(len(ids)):
        if ids[i] not in seen:
            seen.append(ids[i])
        else:
            seen = [ids[i]]

        if len(seen) > max_len:
            max_len = len(seen)

    return max_len

ids = list(map(int, input("Enter student IDs: ").split()))
print("Longest unique sequence:", longest_attendance(ids))
```

```python
def max_discount(arr):
    current = 0
    maximum = arr[0]

    for num in arr:
        current = current + num

        if current > maximum:
            maximum = current

        if current < 0:
            current = 0

    return maximum


arr = list(map(int, input("Enter discounts: ").split()))
print("Maximum discount:", max_discount(arr))
```

```python
def trap_water(height):
    water=0

    for i in range(1, len(height)-1):
        left=max(height[:i])
        right=max(height[i+1:])

        minimum=min(left, right)

        if minimum>height[i]:
            water+=minimum-height[i]

    return water


height = list(map(int, input("Enter building heights: ").split()))
print("Total trapped water:", trap_water(height))
```

```python
def best_performance(scores):
    current = scores[0]
    maximum = scores[0]

    for i in range(1, len(scores)):
        if current + scores[i] > scores[i]:
            current += scores[i]
        else:
            current = scores[i]

        if current > maximum:
            maximum = current

    return maximum

scores = list(map(int, input("Enter performance scores: ").split()))
print("Best performance:", best_performance(scores))
```

```python
def max_product(nums):
    maximum = nums[0]
    minimum = nums[0]
    answer = nums[0]

    for i in range(1, len(nums)):
        if nums[i] < 0:
            maximum, minimum = minimum, maximum

        maximum = max(nums[i], maximum * nums[i])
        minimum = min(nums[i], minimum * nums[i])

        if maximum > answer:
            answer = maximum

    return answer

nums = list(map(int, input("Enter sales values: ").split()))
print("Maximum product:", max_product(nums))
```

```python
def unique_products(products):
    left = 0
    seen = []
    maximum = 0

    for right in range(len(products)):
        while products[right] in seen:
            seen.remove(products[left])
            left += 1

        seen.append(products[right])

        if len(seen) > maximum:
            maximum = len(seen)

    return maximum

products = input("Enter product IDs: ").split()
print("Longest unique purchase sequence:", unique_products(products))
```

```python
def count_transactions(nums, target):
    count = 0

    for i in range(len(nums)):
        total = 0

        for j in range(i, len(nums)):
            total += nums[j]

            if total == target:
                count += 1
    return count

nums = list(map(int, input("Enter transactions: ").split()))
target = int(input("Enter target: "))

print("Number of groups:", count_transactions(nums, target))
```

```python
def group_skills(words):
    groups = {}

    for word in words:
        key = "".join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())

words = input("Enter words: ").split()

result = group_skills(words)

print("Groups:")
for group in result:
    print(group)
```

```python
def longest_sequence(nums):
    numbers = set(nums)
    longest = 0

    for n in numbers:
        if n - 1 not in numbers:
            length = 1

            while n + length in numbers:
                length += 1

            if length > longest:
                longest = length

    return longest

nums = list(map(int, input("Enter packet IDs: ").split()))
print("Longest consecutive sequence:", longest_sequence(nums))
```

```python
def merge_intervals(intervals):
    intervals.sort()
    merged = [intervals[0]]

    for current in intervals[1:]:
        last = merged[-1]

        if current[0] <= last[1]:
            if current[1] > last[1]:
                last[1] = current[1]
        else:
            merged.append(current)

    return merged


n = int(input("Enter number of appointments: "))
intervals = []

for i in range(n):
    start, end = map(int, input(f"Appointment {i+1}: ").split())
    intervals.append([start, end])

result = merge_intervals(intervals)

print("Merged Schedule:")
for item in result:
    print(item)
```
